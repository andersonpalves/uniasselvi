```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# 1. Carregar dados
iris = load_iris()
X = iris.data
y = iris.target
labels = iris.target_names

# 2. Verificar as dimensões originais
print("Formato original:", X.shape)  # (150, 4) → 150 amostras, 4 dimensões

# 3. Aplicar PCA para reduzir de 4 -> 2 dimensões
pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X)

# 4. Visualizar resultado
plt.figure(figsize=(8,6))
for i, label in enumerate(labels):
    plt.scatter(X_reduced[y == i, 0], X_reduced[y == i, 1], label=label)

plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")
plt.title("Redução de Dimensionalidade com PCA")
plt.legend()
plt.show()
```
